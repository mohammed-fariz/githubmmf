```python
from ultralytics import YOLO
import shutil

model = YOLO("yolov8m.pt")

model.train(
    data="data.yaml",
    epochs=50,
    imgsz=640,
    batch=16,

    # Learning
    lr0=0.001,
    lrf=0.01,

    # Augmentation (balanced)
    augment=True,
    mosaic=0.8,
    mixup=0.1,
    flipud=0.5,
    fliplr=0.5,
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,
    degrees=10.0,
    translate=0.1,
    scale=0.3,
    shear=2.0,

    # Regularization
    weight_decay=0.0005,

    # Training control
    patience=10,

    # Save
    project=r"C:\Users\T074\git clone\project\Deep Learning\YOLO\yolo detection",
    name="yolov8m_transfer_aug",
    exist_ok=True,
    save=True,
    save_period=5
)

# Copy best model
shutil.copy(
    r"C:\Users\T074\git clone\project\Deep Learning\YOLO\yolo detection\yolov8m_transfer_aug\weights\best_detection.pt"
    "YOLO\yolo detection\yolov8m_transfer_aug\weights\best_detection.pt",
)

print("Best model saved at:")
```

```python
# Validation 


from ultralytics import YOLO

# Load best fine-tuned model
model = YOLO(r"C:\Users\T074\git clone\project\Deep Learning\YOLO\yolo detection\best_detection.pt")

# Run validation on val set
metrics = model.val(data="data.yaml", split="val")

print(metrics)
```
