```python
#TRANSFER LEARNING WITH AUGMENTATION + REGULARIZATION




from ultralytics import YOLO

# ==============================
# LOAD PRETRAINED YOLOv8m
# ==============================
model = YOLO("yolov8m.pt")

# ==============================
# TRAIN WITH AUGMENTATION + REGULARIZATION
# ==============================
model.train(
    data=r"/content/drive/MyDrive/computer_vision/truck/data.yaml",
    epochs=50,
    imgsz=640,
    batch=16,

    # Learning settings
    lr0=0.001,
    lrf=0.01,

    # AUGMENTATION (VERY IMPORTANT)
    augment=True,
    mosaic=1.0,        # combine 4 images
    mixup=0.2,         # blend images
    flipud=0.5,        # vertical flip
    fliplr=0.5,        # horizontal flip
    hsv_h=0.015,       # color variation
    hsv_s=0.7,
    hsv_v=0.4,
    degrees=10.0,      # rotation
    translate=0.1,
    scale=0.5,
    shear=2.0,

    #  REGULARIZATION
    weight_decay=0.0005,   # L2 regularization
    dropout=0.1,           # dropout (YOLOv8 supports partially)

       #  SAVE SETTINGS
    project="/content/drive/MyDrive/computer_vision/truck",
    name="transfer_aug",
    save=True,
    save_period=5      # save every 5 epochs
)

# ==============================
# BEST MODEL PATH
# ==============================
print("Best model saved at:")
import shutil

shutil.copy(
    "/content/drive/MyDrive/computer_vision/truck/transfer_aug/weights/best.pt",
    "/content/drive/MyDrive/computer_vision/truck/best_detection.pt"              # rename here
)

print("/content/drive/MyDrive/computer_vision/truck/best_detection.pt")
```

```python
# Validation


from ultralytics import YOLO

# Load best fine-tuned model
model = YOLO(r"/content/drive/MyDrive/computer_vision/truck/best_detection.pt")

# Run validation on val set
metrics = model.val(
    data=r"/content/drive/MyDrive/computer_vision/truck/data.yaml",
    split="val"
)



print(metrics)
```
